### Step 2: three buildings, and one prediction

Three example buildings, rounded from real ones in our table, all with the same wall U-value (1.5),
so only compactness differs. As in session 1 the model is a line: predicted demand = w × compactness
+ b, with the weight w and the bias b that lifts the whole line. With w = 8 and b = 110 the first
building, compactness 4.0, gets 8 × 4.0 + 110 = 142 kWh/m²a.

**Guess first:** what does the line predict for the other two?

```python
# Compactness of the three buildings
compactness = [4.0, 6.0, 8.0]
# Their heating demand in kWh/m²a (rounded from the table)
demand = [147.0, 165.0, 183.0]

# The model from session 1, a line: weight w times compactness c,
# plus the bias b
def predict(c, w, b):
    return w * c + b

# A first guess for w and b
w = 8
b = 110

# An empty list; append adds one value at its end
predictions = []
# One prediction per building
for c in compactness:
    p = predict(c, w, b)
    predictions.append(p)
    print(f"compactness {c}: predicted {p}")
```

### Step 3: the MSE by foot, and your part

How wrong is the line, as one number? For each building: the miss (predicted minus actual), the miss
squared, then all squares added up and divided by how many. The loop below keeps a running total,
like Check 1. **Guess first:** what will the total be after all three buildings?

One line is yours: add the square to the running total. In Python, "total becomes total plus square"
reads `total = total + square`.

```python
# The running total of squared misses, starting at 0
total = 0
# range(3) gives the positions 0, 1, 2: one per building
for n in range(3):
    # The prediction for building n, and its miss (predicted - actual)
    p = predict(compactness[n], 8, 110)
    miss = p - demand[n]
    # The miss squared: always positive, big misses count more
    square = miss * miss

    # YOUR TURN: replace the line "raise ..." with one line:
    #   total becomes total plus square
    raise NotImplementedError("replace this line with the line above")

    print(f"building {n}: miss {miss}, squared {square}, total so far {total}")

# Divide by how many buildings: the mean
count = len(demand)
mse_by_hand = total / count
print(f"MSE = {total} / {count} = {mse_by_hand:.2f}")
```

<details>
<summary>Answer</summary>

    total = total + square

The misses are −5, −7 and −9, their squares 25, 49 and 81, the total 155, divided by 3: 51.67.

</details>

We need the same computation for other values of w and b, so the next cell packs exactly this loop
into a function, `mse(w, b)`. Nothing new in it.

```python
def mse(w, b):
    """Mean squared error of the line w * c + b on our three buildings.
    (Text in triple quotes right under def describes the function.)"""
    # The same running total as above. The function uses the lists
    # compactness and demand from the cell of step 2.
    total = 0
    for n in range(3):
        p = predict(compactness[n], w, b)
        miss = p - demand[n]
        square = miss * miss
        total = total + square
    # Hand back the mean
    count = len(demand)
    return total / count


# Check: the same as the loop above
check = mse(8, 110)
print(f"mse(8, 110) = {check:.2f}")
```

### Step 4: the gradient by foot, with a tiny step

Which way should w go? Make w a tiny bit bigger, 0.001, and see how the MSE changes. The change
divided by 0.001 is an **estimate of the gradient** for w: how fast the MSE grows per unit of w.
Negative means: making w bigger makes the error smaller.

**Guess first:** will the gradient for w come out positive or negative?

```python
# The size of the tiny step
tiny = 0.001
# The MSE where we stand: w = 8, b = 110
here = mse(8, 110)
# The MSE with w a tiny step bigger
further = mse(8 + tiny, 110)
# How much the MSE changed
change = further - here
# Change per unit of w: the gradient estimate
gradient_w = change / tiny

print(f"MSE here:               {here:.6f}")
print(f"MSE with w + 0.001:     {further:.6f}")
print(f"change:                 {change:.6f}")
print(f"change / 0.001:         {gradient_w:.2f}")
```

<details>
<summary>Answer</summary>

Negative, about −89.3: the MSE falls by about 0.089 when w grows by 0.001. So w should grow. It is
large because w is multiplied by compactness, 4 to 8, so moving w moves every prediction a lot.

</details>

The same for b, and both packed into one function for the learning loop. Both gradients are computed
**before** either number changes; otherwise the second would be measured on a line that has already
moved.

```python
def gradient_estimates(w, b):
    """Both gradients of the MSE, each estimated with a tiny step."""
    tiny = 0.001
    here = mse(w, b)
    # A tiny step in w, b unchanged
    further_w = mse(w + tiny, b)
    # A tiny step in b, w unchanged
    further_b = mse(w, b + tiny)
    # Change per unit, for each
    g_w = (further_w - here) / tiny
    g_b = (further_b - here) / tiny
    # Hand back two values at once, in this order: g_w, then g_b
    return g_w, g_b


# Catch the two values in two names, in the same order
g_w, g_b = gradient_estimates(8, 110)
print(f"gradient for w: {g_w:.2f}")
print(f"gradient for b: {g_b:.2f}")
```

### Step 5: two steps of the learning loop, with all the arithmetic

One step: w minus learning rate times its gradient, b minus learning rate times its gradient. The
minus is what makes the step go **against** the gradient, downhill. Learning rate 0.01.

**Guess first:** after the first step, will w be bigger or smaller than 8? Will the MSE be lower?

```python
# Start from our first guess
w = 8
b = 110
# How big the steps are
learning_rate = 0.01

start_mse = mse(w, b)
print(f"start: w = {w}, b = {b}, MSE = {start_mse:.2f}")
# Two steps, every number printed
for step in [1, 2]:
    # Both gradients first, where we stand now
    g_w, g_b = gradient_estimates(w, b)
    # Then both updates: minus, so the step goes against the gradient
    new_w = w - learning_rate * g_w
    new_b = b - learning_rate * g_b
    print(f"step {step}:")
    print(f"  w = {w:.3f} - {learning_rate} * ({g_w:.2f}) = {new_w:.3f}")
    print(f"  b = {b:.3f} - {learning_rate} * ({g_b:.2f}) = {new_b:.3f}")
    # The new numbers become the current ones
    w = new_w
    b = new_b
    current_mse = mse(w, b)
    print(f"  MSE now {current_mse:.2f}")
```

<details>
<summary>Answer</summary>

Bigger: 8 − 0.01 × (−89.29) = 8.893. The MSE drops from 51.67 to 2.29. The second step pushes w
slightly past 9, to 9.079, and the MSE drops to 0.15.

</details>

### Step 6: two inputs, and the kink, by foot

With two inputs, each gets its own weight, and the prediction is a **weighted sum**: weight 1 times
input 1, plus weight 2 times input 2, plus the bias. Session 2 found the weights 9 and 62 and the
bias 18. **Guess first:** what does the weighted sum give for the first building, compactness 4.0
and U-value 1.5?

```python
# The second input: the U-value of each building
u_value = [1.5, 1.5, 1.5]

# The formula session 2 found: 9 * compactness + 62 * U-value + 18
for n in range(3):
    part_1 = 9 * compactness[n]
    part_2 = 62 * u_value[n]
    weighted_sum = part_1 + part_2 + 18
    print(f"building {n}: {part_1} + {part_2} + 18 = {weighted_sum}")
```

<details>
<summary>Answer</summary>

9 × 4.0 + 62 × 1.5 + 18 = 36 + 93 + 18 = 147, exactly the table. The other two give 165 and 183:
these demands were computed with that formula in the first place.

</details>

The kink from session 2, **ReLU**, is just an `if`: a negative number becomes 0, anything else stays.
Below, two ramps, one with its kink at 5 and one at 7, added up with weights 2 and 3. **Guess
first:** what does the sum give at compactness 8?

```python
# ReLU: keep a number if it is positive, otherwise 0
def relu(z):
    # Below zero: hand back 0, and the function stops here
    if z < 0:
        return 0.0
    # Only reached when z is not below zero: hand back z itself
    return z

# A ramp with its kink at 5, one with its kink at 7, added up
for c in [4.0, 5.0, 6.0, 7.0, 8.0]:
    ramp_1 = relu(c - 5)
    ramp_2 = relu(c - 7)
    together = 2 * ramp_1 + 3 * ramp_2
    print(f"compactness {c}: ramps {ramp_1} and {ramp_2}, sum {together}")
```

<details>
<summary>Answer</summary>

At 8: the first ramp gives 8 − 5 = 3, the second 8 − 7 = 1, so 2 × 3 + 3 × 1 = 9. Up to 5 the sum is
0; from 5 on it rises by 2 per unit, from 7 on by 2 + 3 = 5 per unit. Two ramps already make a line
with two kinks. A network is many of these, added up with weights.

</details>

### Optional, if time allows: the full learning loop

The learning loop of session 1 without any shortcut: 2000 steps, printing every 500th.

```python
# Start again from the first guess
w = 8
b = 110
learning_rate = 0.01

# range(1, 2001) counts 1, 2, ... 2000 (the end is left out)
for step in range(1, 2001):
    # Both gradients, then both updates
    g_w, g_b = gradient_estimates(w, b)
    w = w - learning_rate * g_w
    b = b - learning_rate * g_b
    # Print only every 500th step: step % 500 is what is left over
    # when step is divided by 500, and that is 0 every 500 steps
    if step % 500 == 0:
        current_mse = mse(w, b)
        print(f"step {step}: w = {w:.3f}, b = {b:.3f}, MSE = {current_mse:.4f}")
```

<details>
<summary>Where is it heading?</summary>

Close to w = 9 and b = 111, where the line goes exactly through the three buildings. It does not get
there exactly, because the tiny step only estimates the gradient; it ends near w = 9.00 and b = 111.0,
with an MSE that rounds to 0.0000 but is not quite 0. b moves slowly: the crawl from session 2,
because compactness is not rescaled.

</details>

## Part 2: the same with numpy and pandas, as shortcuts (10 minutes)

**numpy** gives you lists you can compute with in one go: an **array**. Three kinds of shortcut,
each next to its loop:

- **element by element**: `*`, `+`, `-`, `**` on an array work on every element separately;
- **weighted sums**: `@` multiplies each row of a table with the weights and adds up;
- **summaries**: `np.mean(...)` or `.mean(axis=0)` turn many numbers into one per column.

**Guess first:** what is `[1, 2, 3] * 2` for a plain list?

```python
# "import numpy as np": load numpy and call it np for short
import numpy as np

# A plain list times 2: the list is REPEATED
doubled_list = [1, 2, 3] * 2
print("list * 2: ", doubled_list)
# An array times 2: every ELEMENT is multiplied
doubled_array = np.array([1, 2, 3]) * 2
print("array * 2:", doubled_array)

# Our three buildings as arrays
c = np.array(compactness)
d = np.array(demand)

# Element by element: all three predictions at once
p = 8 * c + 110
print("predictions with numpy:", p)
# The same list, made with the loop in step 2
print("predictions by foot:   ", predictions)
```

<summary>Answer</summary>

`[1, 2, 3, 1, 2, 3]`: on a plain list, `*` repeats it. On an array it multiplies every element. This
is why the code in sessions 1 to 3 always turns lists into arrays first.

</details>

The MSE of step 3, as shortcuts. **Guess first:** will it be the same number as your loop?

<details>
<summary>Answer</summary>

</details>

Weighted sums and column averages. `X` is a table: one row per building, one column per input,
written as a list of rows, hence the double square brackets. **Guess first:** what is the average of
the second column, the U-values?

```python
# A table: each inner list is one row (one building):
# [compactness, U-value]
X = np.array([[4.0, 1.5],
              [6.0, 1.5],
              [8.0, 1.5]])
# The two weights, one per column
weights = np.array([9, 62])

# By foot: for every row, weight times input, added up, plus 18
for row in X:
    by_foot = row[0] * 9 + row[1] * 62 + 18
    print("weighted sum by foot:", by_foot)
# Shortcut: @ does that weighted sum for every row at once
with_at = X @ weights + 18
print("weighted sums with @: ", with_at)

# By foot: the average of each column
average_c = (4.0 + 6.0 + 8.0) / 3
average_u = (1.5 + 1.5 + 1.5) / 3
print("column averages by foot:", average_c, average_u)
# Shortcut: axis=0 means "go down the columns"
with_axis = X.mean(axis=0)
print("column averages with numpy:", with_axis)
```

<details>
<summary>Answer</summary>

1.5: all three are 1.5. `X @ weights` is not element by element: it combines each row into one
number. `X.mean(axis=0)` is not element by element either: it summarises each column into one number.

</details>
